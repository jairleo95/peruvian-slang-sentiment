```python
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
```

```python

# Ejemplo de datos para tres categorías

#roBERTuito
y_true1 = [0, 1, 2, 2, 1, 0]
y_pred1 = [0, 1, 2, 1, 1, 0]

#BETO
y_true2 = [0, 1, 1, 2, 0, 2]
y_pred2 = [0, 2, 1, 2, 0, 1]


y_true3 = [2, 0, 1, 0, 2, 1]
y_pred3 = [2, 0, 1, 0, 2, 1]

y_true4 = [0, 2, 0, 1, 1, 2]
y_pred4 = [1, 2, 0, 1, 0, 2]

# Crear las matrices de confusión
cm1 = confusion_matrix(y_true1, y_pred1)
cm2 = confusion_matrix(y_true2, y_pred2)
cm3 = confusion_matrix(y_true3, y_pred3)
cm4 = confusion_matrix(y_true4, y_pred4)

# Configurar la figura y los subplots
fig, axs = plt.subplots(2, 2, figsize=(10, 10))

# Mostrar cada matriz de confusión en un subplot
labels = ['Negative', 'Neutral', 'Positive']

ConfusionMatrixDisplay(cm1, display_labels=labels).plot(ax=axs[0, 0])
axs[0, 0].set_title('Matriz de Confusión 1')

ConfusionMatrixDisplay(cm2, display_labels=labels).plot(ax=axs[0, 1])
axs[0, 1].set_title('Matriz de Confusión 2')

ConfusionMatrixDisplay(cm3, display_labels=labels).plot(ax=axs[1, 0])
axs[1, 0].set_title('Matriz de Confusión 3')

ConfusionMatrixDisplay(cm4, display_labels=labels).plot(ax=axs[1, 1])
axs[1, 1].set_title('Matriz de Confusión 4')

# Ajustar el layout
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay
import numpy as np

# Cargar los datos de las matrices de confusión desde archivos CSV
cm1 = pd.read_csv('results/roBERTuito_val_df_confusion_percentages.csv', index_col=0).values
cm2 = pd.read_csv('results/BERTuit_val_df_confusion_percentages.csv', index_col=0).values
cm3 = pd.read_csv('results/MARIA_val_df_confusion_percentages.csv', index_col=0).values
cm4 = pd.read_csv('results/BETO_val_df_confusion_percentages.csv', index_col=0).values

# Configurar las etiquetas de las categorías
labels = ['Negative', 'Neutral', 'Positive']

# Configurar la figura y los subplots
fig, axs = plt.subplots(2, 2, figsize=(12, 12))

# Configurar el estilo
# plt.style.use('seaborn-v0_8-dark-palette')

# Función para mostrar la matriz de confusión con porcentajes
def plot_confusion_matrix(cm, ax, title):
    disp = ConfusionMatrixDisplay(cm, display_labels=labels)
    disp.plot(ax=ax, cmap='viridis', values_format='.2%', colorbar=False)
    ax.set_title(title, fontsize=14)
    ax.set_xlabel('Predicted label', fontsize=12)
    ax.set_ylabel('True label', fontsize=12)
    ax.set_xticklabels(labels, fontsize=10)
    ax.set_yticklabels(labels, fontsize=10)

# Mostrar cada matriz de confusión en un subplot
plot_confusion_matrix(cm1, axs[0, 0], 'RoBERTuito')
plot_confusion_matrix(cm2, axs[0, 1], 'BERTuit')
plot_confusion_matrix(cm3, axs[1, 0], 'RoBERTa Base BNE')
plot_confusion_matrix(cm4, axs[1, 1], 'BETO')

# Ajustar el layout
plt.tight_layout()

plt.savefig('confusion_matrices.png', dpi=300)  # Ajustar dpi para mayor calidad

plt.show()
```

```python
# df = df[df.ANNOTATOR.isin(["0"])]
dfa = dfa[dfa.label.isin(["1","2","3"])]
#new_df = new_df.loc[:, ['id_slang','slang', 'sentence', 'label']]
dfa = dfa.loc[:, ['text', 'label']]
dfa = dfa.reset_index(drop=True)
dfa = dfa.astype({'label':'int'})
dfa
```

```python
dfa.info()
```

```python
# label encode
def label_encode(x):
    if x == 1:#positive
        return 2
    if x == 2:#neutral
        return 1
    if x == 3:#negative
        return 0
# label to name
def label2name(x):
    if x == 0:
        return "negative"
    if x == 1:
        return "neutral"
    if x == 2:
        return "positive"
```

```python
dfa["label"] = dfa["label"].apply(lambda x: label_encode(x))
```

## Load Raw Dataset

```python
# merging two csv files 
dataset_path = '../data/'
df = pd.concat( 
    map(pd.read_csv, [dataset_path + 'raw_dataset-01.csv',
                      dataset_path + 'raw_dataset-02.csv',
                      dataset_path + 'raw_dataset-03.csv',
                      dataset_path + 'raw_dataset-04.csv',
                      dataset_path + 'raw_dataset-05.csv',
                      dataset_path + 'raw_dataset-06.csv']), ignore_index=True)
# print(df) 
df.info()
```

```python
df["label_name"] = df["label"].apply(lambda x: label2name(x))
```

```python
df.sample(15)
```

# Split Train, Test and Validaiton

```python

from sklearn.model_selection import train_test_split


random_state=1
train_df_, val_df = train_test_split(df, 
                                    test_size=0.20, 
                                    random_state=random_state, 
                            stratify=df.label.values)
```

```python

train_df, test_df = train_test_split(train_df_, 
                                    test_size=0.20, 
                                    random_state=42, 
                            stratify=train_df_.label.values)
```
